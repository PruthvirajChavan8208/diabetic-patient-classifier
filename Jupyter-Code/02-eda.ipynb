```python
ds.columns
X = ds[['Pregnancies', 'Glucose', 'BloodPressure','SkinThickness', 'Insulin', 'BMI', 'DiabetesPedigreeFunction', 'Age']]
y = ds[["Outcome"]]
```

```python
#Lets increase samples for class 1
from imblearn.over_sampling import SMOTE
from collections import Counter
smote = SMOTE(random_state=42)
X, y = smote.fit_resample(X, y)
```

```python
len(X)
```

```python
len(y)
```

```python
ds = pd.concat([X, y], axis = 1)
```

```python
ds.to_csv("cleaned_data.csv")
```
